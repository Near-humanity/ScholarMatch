```python
import pandas as pd

df = pd.read_csv("papers.csv")

recent_papers = df[df["year"] >= 2024]
print(recent_papers)

papers_count = recent_papers.shape[0]

print(f"共有{papers_count}篇论文")

sorted_papers = df.sort_values("year", ascending=True)
print(sorted_papers)
```

## 2026.9.17
- pd 就是pandas的缩写，方便输入
- df means DataFrame（数据框），是给 pd.read_csv() 返回的表格起的变量名，属于一种约定俗成

对比：
import csv
with open("papers.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    papers = list(reader)

和

import pandas as pd
df = pd.read_csv("papers.csv")

pandas的优势就是把“打开文件、指定编码、读成字典、转成列表”这四步合并成了一步。

1. 筛选：recent_papers = df[df["year"] >= 2024] —— 用于从大量论文里挑出符合时间条件的
2. 计数：papers_count = len(筛选后的变量名) 或者 筛选后的变量名.shape[0] —— 用于统计筛选后的论文数量
3. 排序：sort_values("year", ascending=False), 对df中某一列的数值进行排序函数是sort_values，后面加上“列的表头”，函数默认为从小到大排序，或者加上ascending=True，函数从大到小排序代码语言是ascending=False。  —— 用于把最新发表的论文排在最前面
** sort_index是按“行号”排，一般不用
