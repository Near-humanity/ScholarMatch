"w" = write，写入（会清空原文件）

把这段代码翻译成中文，就是：

导入 csv 工具包。
打开 papers.csv 这个文件，用只读模式，按 UTF-8 编码读。
把这个文件交给 csv 工具包里的“字典读取器”处理，结果存进 reader。
把 reader 里的所有行读出来，转成一个列表，存进 papers。

```python
import csv
with open("papers.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    papers = list(reader)

def count_by_journal(papers):
    count = {}
    for paper in papers:
        journal = paper["journal"]
        if journal in count:
            count[journal] += 1
        else:
            count[journal] = 1
    return count

journal_count = count_by_journal(papers)
print(journal_count)
print(journal_count.keys())     # 只看键，也就是所有期刊名
print(journal_count.values())   # 只看值，也就是所有计数
print(journal_count.items())    # 同时看键和值

for journal, n in journal_count.items():
    print(f"{journal}: {n}")
```

方括号 [] = 列表
列表是“一排格子”，里面装的东西是按顺序排列的。你只关心“第一个是什么、第二个是什么”，不关心“它叫什么名字”。

花括号 {} = 字典
字典是“一张卡片”，里面装的是键—值对。你关心的是“这个键对应什么值”。


括号	类型	装什么	怎么访问	什么时候用
[]	列表 list	按顺序排列的一堆东西	result[0]	筛选、排序、保存一串结果
{}	字典 dict	键—值对	journal_count["键"]	计数、映射、查找对应关系


filter_by_year 的任务是： “从一堆论文里，挑出符合条件的那些。”
它只需要“把挑出来的论文排成一排”，不需要给每篇论文起名字。
所以用列表 result = []。

count_by_journal 的任务是： “统计每个期刊出现了多少次。”
它需要知道“J. Med. Chem. 出现了 1 次、Scientometrics 出现了 1 次”。
期刊名就是“键”，次数就是“值”。
所以用字典 journal_count = {}。

.items() 是字典的一个方法，它把字典里的每一对“键—值”拿出来，变成一个个元组。
然后 for journal, n in ... 的意思是：每一轮循环，把元组里的第一个元素存进 journal，第二个元素存进 count。
