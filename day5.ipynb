```python
import pandas as pd

df = pd.read_csv("Latest_Organic_Chemistry_Papers.csv", encoding="utf-8-sig")

import ast
from collections import Counter

df["keywords_list"] = df["keywords"].apply(
    lambda x: ast.literal_eval(x) if isinstance(x, str) else []
)

all_keywords = []
for kw_list in df["keywords_list"]:
    all_keywords.extend(kw_list)

counter = Counter(all_keywords)
top_keywords = pd.DataFrame(counter.most_common(20), columns=["keyword", "count"])
print(top_keywords)
```
